# Arrow-of-time symmetry controls for frozen V-JEPA 2 (exploratory)

Runs `scripts/explore_arrow_symmetry_vjepa2.py` from a clone of the public repository at a fixed commit, on a Kaggle GPU (T4 or P100, internet on). Download `arrow-symmetry-vjepa2.json` from the Output panel; it goes to `results/exploratory/`. Expected runtime: about 15 min of simulation on CPU, then about 40 min of encoding on a T4.


```python
# ---- settings ---------------------------------------------------------------
GIT_COMMIT = "<full 40-character commit SHA>"  # the commit to install and measure
TRANSFORMERS_VERSION = "5.24.1"
DRY_RUN = False  # True: smoke profile and a tiny random V-JEPA 2 on CPU, to test the notebook
```

```python
import re
import subprocess
import sys

assert re.fullmatch(r"[0-9a-f]{40}", GIT_COMMIT), "GIT_COMMIT must be a full 40-character SHA"
url = "https://github.com/gabrielkpo/weather-world-model"
src = "/kaggle/working/wwm-src"
# --no-deps: numpy and torch are already in the Kaggle image (CUDA build).
pip = [sys.executable, "-m", "pip", "install", "-q", "--ignore-requires-python"]
subprocess.run([*pip, "--no-deps", "--force-reinstall", f"git+{url}@{GIT_COMMIT}"], check=True)
subprocess.run([*pip, f"transformers=={TRANSFORMERS_VERSION}"], check=True)
subprocess.run(["git", "clone", "-q", url, src], check=True)
subprocess.run(["git", "-C", src, "checkout", "-q", GIT_COMMIT], check=True)
```

```python
script = f"{src}/scripts/explore_arrow_symmetry_vjepa2.py"
cmd = [sys.executable, script, "--out", "/kaggle/working"]
subprocess.run(cmd + (["--dry-run"] if DRY_RUN else []), cwd=src, check=True)
```
